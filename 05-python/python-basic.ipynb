```python
midtern, finals, quiz = 99, 77, 33
print(midtern)
print(finals)
print(quiz)

result = midtern + finals + quiz
print(f'result : {result} ')
print("------------------------ 실습 --------------------------")

result = midtern * 0.4 + finals * 0.4 + quiz * 0.2
print(f'result : {result} ')
print("-------------------자료형 확인 실습---------------------")
#int

print(type(midtern))
print(type(finals))
print(type(quiz))

#float
print(type(result))

print("------------Bloolean 자료형 확인 실습--------------")

t = True
f = False
print(type(t))
print(type(f))

#논리 연산자
print("-----------논리 연산자 자료형 확인 실습-------------")

print(t and f)
print(t or f)
print(not t)

## 비교 연산자

print(t == f)

print(1!=2)
```

```python
#반복문 for - 조직을 만드는 것은 다 할 수 있음

for i in range(5):
  print(i)

for i in range(5):
  print('*', end=' ')

print(end='\n')

fruits = ['apple', 'banana', 'cherry']

for i in fruits:

  print(i)

for index, value in enumerate(fruits):
  print(index, value)

for index in range(len(fruits)):
  print(index, fruits[index])
```
